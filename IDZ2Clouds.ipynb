```python
from flask import Flask, request, jsonify, render_template_string
from transformers import pipeline
from pyngrok import ngrok
import threading

sentiment_pipeline = pipeline("sentiment-analysis")

app = Flask(__name__)

HTML_TEMPLATE = """
<!DOCTYPE html>
<html>
<head><title>AI Sentiment Analyzer</title></head>
<body style="font-family: Arial; margin: 40px;">
    <h2>Аналіз тональності тексту (Google Colab + AI)</h2>
    <form method="POST">
        <textarea name="text" rows="4" cols="50" placeholder="Введіть текст англійською..."></textarea><br><br>
        <button type="submit">Аналізувати</button>
    </form>
    {% if result %}
        <h3>Результат аналізу:</h3>
        <table border="1" cellpadding="8" style="border-collapse: collapse;">
            <tr><th>Мітка (Label)</th><th>Впевненість (Score)</th></tr>
            <tr><td>{{ result[0]['label'] }}</td><td>{{ "%.4f"|format(result[0]['score']) }}</td></tr>
        </table>
    {% endif %}
</body>
</html>
"""

@app.route("/", methods=["GET", "POST"])
def home():
    result = None
    if request.method == "POST":
        user_text = request.form.get("text")
        if user_text:
            result = sentiment_pipeline(user_text)
    return render_template_string(HTML_TEMPLATE, result=result)

NGROK_TOKEN = "3Lr5mi1MIfl1mwVMu1760RzbNlY_3VMvgpRVujm4YdKrP28d9"
ngrok.set_auth_token(NGROK_TOKEN)

public_url = ngrok.connect(5000)
print("Публічна URL-адреса:", public_url)

threading.Thread(target=app.run, kwargs={"port": 5000}).start()
```
